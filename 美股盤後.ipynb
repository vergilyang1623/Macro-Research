Basic helpers

```python
# ---------- 基本工具 ----------
import base64
import html
import io
import re
from concurrent.futures import ThreadPoolExecutor
from dataclasses import dataclass
from datetime import datetime, timedelta, timezone
from typing import Any, Callable, Optional
from urllib.parse import quote_plus
from zoneinfo import ZoneInfo

import feedparser
import pandas as pd
import requests
import yfinance as yf
from IPython.display import Markdown, display

TZ_LOCAL = ZoneInfo("Asia/Taipei")
TZ_ET = ZoneInfo("America/New_York")
HOUR_LABEL = {"bmo": "盤前", "amc": "盤後", "dmh": "盤中"}
HEADERS = {
    "User-Agent": ("Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                   "(KHTML, like Gecko) Chrome/124.0 Safari/537.36"),
    "Accept": "application/rss+xml, application/xml, application/json, text/xml, */*",
}


@dataclass
class Item:
    title: str
    url: str
    source: str
    published: Optional[datetime]
    summary: str = ""


@dataclass
class Task:
    section: str            # 新聞："market" / "ticker" / "official"；數據："data"
    group: Optional[str]    # 個股代號，或數據的名稱
    label: str              # 資料來源狀況表上的名稱
    fn: Callable[[], Any]


def get(url, headers=None, params=None):
    h = dict(HEADERS)
    if headers:
        h.update(headers)
    r = requests.get(url, headers=h, params=params, timeout=TIMEOUT)
    r.raise_for_status()
    return r


def clean_text(s, limit=180):
    if not s:
        return ""
    s = html.unescape(re.sub(r"<[^>]+>", " ", s))
    s = re.sub(r"\s+", " ", s).strip()
    return s if len(s) <= limit else s[:limit].rstrip() + "…"


def entry_time(entry):
    t = entry.get("published_parsed") or entry.get("updated_parsed")
    return datetime(*t[:6], tzinfo=timezone.utc) if t else None


def norm_key(title):
    return re.sub(r"[\W_]+", "", title.lower())[:80]  # 保留中文字，中英文標題都能去重


def is_key(title):
    acr = r"\b(?:" + "|".join(map(re.escape, ACRONYMS)) + r")\b"
    kw = r"\b(?:" + "|".join(map(re.escape, KEYWORDS)) + r")\w*"
    return bool(re.search(acr, title) or re.search(kw, title, re.I))


def pct(x, digits=2):
    return "—" if x is None or pd.isna(x) else f"{x:+.{digits}f}%"


def num(x, digits=2):
    return "—" if x is None or pd.isna(x) else f"{x:,.{digits}f}"


def billions(x):
    return f"{x / 1e9:,.2f}B" if x else "—"


def md_table(headers, rows):
    cell = lambda c: str(c).replace("|", "/")
    lines = ["| " + " | ".join(headers) + " |", "|" + "|".join(["---"] * len(headers)) + "|"]
    lines += ["| " + " | ".join(cell(c) for c in r) + " |" for r in rows]
    return "\n".join(lines)


def safe_err(e):
    """錯誤訊息去掉金鑰後再顯示"""
    msg = re.sub(r"\s+", " ", f"{type(e).__name__}: {e}")
    for k in (globals().get("FINNHUB_API_KEY"), globals().get("FRED_API_KEY")):
        if k:
            msg = msg.replace(k, "***")
    return msg[:160]


def et_today():
    return datetime.now(TZ_ET).date()


def next_weekday(d):
    d += timedelta(days=1)
    while d.weekday() >= 5:
        d += timedelta(days=1)
    return d


def dedup_sort(items, limit):
    seen, out = set(), []
    for it in items:
        k = norm_key(it.title)
        if not k or k in seen or (it.url and it.url in seen):
            continue
        seen.add(k)
        if it.url:
            seen.add(it.url)
        out.append(it)
    out.sort(key=lambda x: x.published, reverse=True)
    return out[:limit]


def fmt(it, show_summary):
    t = it.published.astimezone(TZ_LOCAL).strftime("%m/%d %H:%M")
    star = "★ " if is_key(it.title) else ""
    title = it.title.replace("[", "(").replace("]", ")")
    line = f"- `{t}` {star}[{title}]({it.url}) — *{it.source}*"
    if show_summary and it.summary and norm_key(it.summary)[:60] != norm_key(it.title)[:60]:
        line += f"\n  > {it.summary}"
    return line


def news_list(items, empty="- （無）"):
    return "\n".join(fmt(it, SHOW_SUMMARY) for it in items) or empty
```

News sources

```python
# ---------- 新聞來源 ----------
def parse_feed(url, headers=None):
    feed = feedparser.parse(get(url, headers=headers).content)
    if not feed.entries and not feed.get("version"):
        raise ValueError("回應不是 RSS/Atom（可能被網站阻擋或網址已失效）")
    return feed


def rss_items(url, source, headers=None):
    items = []
    for e in parse_feed(url, headers).entries:
        title = clean_text(e.get("title", ""), 300)
        if title:
            items.append(Item(title, e.get("link", ""), source, entry_time(e), clean_text(e.get("summary", ""))))
    return items


def google_news(query):
    url = ("https://news.google.com/rss/search?q=" + quote_plus(f"{query} when:1d")
           + "&hl=en-US&gl=US&ceid=US:en")
    items = []
    for e in parse_feed(url).entries:
        publisher = (e.get("source") or {}).get("title", "")
        title = clean_text(e.get("title", ""), 300)
        if publisher and title.endswith(" - " + publisher):
            title = title[: -(len(publisher) + 3)]
        if title:
            items.append(Item(title, e.get("link", ""), f"Google News / {publisher or '未知'}", entry_time(e)))
    return items


def sec_8k(ticker):
    url = ("https://www.sec.gov/cgi-bin/browse-edgar?action=getcompany"
           f"&CIK={ticker}&type=8-K&dateb=&owner=include&count=10&output=atom")
    items = rss_items(url, "SEC EDGAR", headers={"User-Agent": SEC_CONTACT})
    for it in items:  # "8-K - APPLE INC (0000320193) (Filer)" → "【8-K】APPLE INC (0000320193)"
        name = re.sub(r"^8-K(/A)?\s*-\s*", "", it.title)
        it.title = "【8-K】" + re.sub(r"\s*\((Filer|Subject)\)\s*$", "", name)
    return items


def _finnhub_items(data):
    if isinstance(data, dict):  # 金鑰錯誤等情況會回傳 {"error": ...}
        raise ValueError(data.get("error", str(data)))
    items = []
    for d in data:
        if d.get("headline"):
            ts = d.get("datetime")
            items.append(Item(clean_text(d["headline"], 300), d.get("url", ""),
                              f"Finnhub / {d.get('source', '')}",
                              datetime.fromtimestamp(ts, tz=timezone.utc) if ts else None,
                              clean_text(d.get("summary", ""))))
    return items


def finnhub_general():
    r = get("https://finnhub.io/api/v1/news", params={"category": "general", "token": FINNHUB_API_KEY})
    return _finnhub_items(r.json())


def finnhub_company(ticker, start, end):
    r = get("https://finnhub.io/api/v1/company-news",
            params={"symbol": ticker, "from": f"{start:%Y-%m-%d}", "to": f"{end:%Y-%m-%d}",
                    "token": FINNHUB_API_KEY})
    return _finnhub_items(r.json())


def build_news_tasks(cutoff, now):
    # 順序即去重時的優先順序：原始來源排在 Google News 前面
    tasks = [Task("market", None, name, lambda u=url, n=name: rss_items(u, n))
             for name, url in MARKET_FEEDS.items()]
    if FINNHUB_API_KEY:
        tasks.append(Task("market", None, "Finnhub 大盤新聞", finnhub_general))
    tasks += [Task("market", None, f"Google News「{q}」", lambda q=q: google_news(q))
              for q in GOOGLE_NEWS_QUERIES]
    for t in [t.upper() for t in TICKERS]:
        tasks.append(Task("ticker", t, f"Yahoo Finance {t}",
                          lambda t=t: rss_items(YAHOO_TICKER_FEED.format(t), "Yahoo Finance")))
        if FINNHUB_API_KEY:
            tasks.append(Task("ticker", t, f"Finnhub {t}", lambda t=t: finnhub_company(t, cutoff, now)))
        if SEC_CONTACT:
            tasks.append(Task("ticker", t, f"SEC 8-K {t}", lambda t=t: sec_8k(t)))
    tasks += [Task("official", None, name, lambda u=url, n=name: rss_items(u, n))
              for name, url in OFFICIAL_FEEDS.items()]
    return tasks
```

Data Sources

```python
# ---------- 數據來源 ----------
def last_two_closes(symbols):
    """回傳 {代號: (最新收盤, 前一收盤, 日期)}"""
    symbols = list(symbols)
    df = yf.download(symbols, period="10d", interval="1d", auto_adjust=False, progress=False)
    closes = df["Close"]
    if isinstance(closes, pd.Series):
        closes = closes.to_frame(symbols[0])
    out = {}
    for s in symbols:
        if s in closes:
            col = closes[s].dropna()
            if len(col) >= 2:
                out[s] = (float(col.iloc[-1]), float(col.iloc[-2]), col.index[-1].date())
    if not out:
        raise RuntimeError("Yahoo Finance 沒有回傳任何行情")
    return out


def screen_quotes(screen_name):
    if not hasattr(yf, "screen"):
        raise RuntimeError("yfinance 版本太舊，請在 Anaconda Prompt 執行 conda update -c conda-forge yfinance")
    try:
        res = yf.screen(screen_name, count=MOVERS_COUNT)
    except TypeError:
        res = yf.screen(screen_name)
    return (res or {}).get("quotes", [])[:MOVERS_COUNT]


def treasury_curve():
    frames, year = [], et_today().year
    for y in (year, year - 1):  # 年初時當年資料可能不到兩筆，再補前一年
        url = ("https://home.treasury.gov/resource-center/data-chart-center/interest-rates/"
               f"daily-treasury-rates.csv/{y}/all?type=daily_treasury_yield_curve"
               f"&field_tdr_date_value={y}&page&_format=csv")
        frames.append(pd.read_csv(io.StringIO(get(url).text)))
        if sum(len(f) for f in frames) >= 2:
            break
    df = pd.concat(frames, ignore_index=True)
    df["Date"] = pd.to_datetime(df["Date"], format="%m/%d/%Y")
    return df.sort_values("Date", ascending=False).reset_index(drop=True)


def fred_api(endpoint, **params):
    """呼叫 FRED 官方 API；出錯時顯示 FRED 回傳的錯誤原因"""
    try:
        r = get(f"https://api.stlouisfed.org/fred/{endpoint}",
                params={"api_key": FRED_API_KEY, "file_type": "json", **params})
    except requests.HTTPError as e:
        resp = getattr(e, "response", None)
        if resp is None:
            raise
        try:
            msg = resp.json().get("error_message", "")
        except Exception:
            msg = ""
        raise ValueError(f"FRED API {resp.status_code}：{msg or resp.reason}") from None
    return r.json()


def fred_series(series_id, start):
    """有 FRED_API_KEY 時用官方 API（較穩定）；沒有時改用公開 CSV 下載"""
    if FRED_API_KEY:
        obs = fred_api("series/observations", series_id=series_id,
                       observation_start=f"{start:%Y-%m-%d}").get("observations", [])
        df = pd.DataFrame(obs, columns=["date", "value"])
    else:
        url = f"https://fred.stlouisfed.org/graph/fredgraph.csv?id={series_id}&cosd={start:%Y-%m-%d}"
        df = pd.read_csv(io.StringIO(get(url).text))
        df.columns = ["date", "value"]
    df["value"] = pd.to_numeric(df["value"], errors="coerce")   # FRED 以 "." 表示缺值
    return df.dropna().reset_index(drop=True)


def fred_calendar(start, days):
    data = fred_api("releases/dates",
                    realtime_start=f"{start:%Y-%m-%d}",
                    realtime_end=f"{start + timedelta(days=days):%Y-%m-%d}",
                    include_release_dates_with_no_data="true", sort_order="asc", limit=1000)
    rows = []
    for d in data.get("release_dates", []):
        name = d.get("release_name", "")
        if any(k.lower() in name.lower() for k in IMPORTANT_RELEASES) and [d["date"], name] not in rows:
            rows.append([d["date"], name])
    return rows


def finnhub_earnings(day):
    r = get("https://finnhub.io/api/v1/calendar/earnings",
            params={"from": f"{day:%Y-%m-%d}", "to": f"{day:%Y-%m-%d}", "token": FINNHUB_API_KEY})
    data = r.json()
    if "earningsCalendar" not in data:
        raise ValueError(str(data.get("error", data))[:100])
    return data["earningsCalendar"]


def yf_watch_earnings(days=14):
    """自選股未來 N 天的財報日 {代號: (日期, EPS 預估, 營收預估)}"""
    out, errors = {}, []
    for t in [t.upper() for t in TICKERS]:
        try:
            cal = yf.Ticker(t).calendar or {}
            d = (cal.get("Earnings Date") or [None])[0]
            if d and et_today() <= d <= et_today() + timedelta(days=days):
                out[t] = (d, cal.get("Earnings Average"), cal.get("Revenue Average"))
        except Exception as e:
            errors.append(e)
    if TICKERS and len(errors) == len(TICKERS):
        raise errors[-1]
    return out


def fear_greed():
    r = get("https://production.dataviz.cnn.io/index/fearandgreed/graphdata",
            headers={"Referer": "https://www.cnn.com/", "Origin": "https://www.cnn.com"})
    return r.json()["fear_and_greed"]


def sp500_constituents():
    """維基百科 S&P 500 成分股：代號、名稱、GICS 類股、子產業"""
    text = get("https://en.wikipedia.org/wiki/List_of_S%26P_500_companies").text
    tables = pd.read_html(io.StringIO(text))
    df = next(t for t in tables if {"Symbol", "GICS Sector"} <= set(t.columns))
    df = df.rename(columns={"Symbol": "symbol", "Security": "name", "GICS Sector": "sector",
                            "GICS Sub-Industry": "industry"})
    df["symbol"] = df["symbol"].astype(str).str.replace(".", "-", regex=False)  # BRK.B → BRK-B（Yahoo 格式）
    return df[["symbol", "name", "sector", "industry"]]


def _shares(symbol):
    try:
        return float(yf.Ticker(symbol).fast_info["shares"])
    except Exception:
        return None


def sp500_universe():
    """成分股 + 流通股數；存成快取檔，SP500_CACHE_DAYS 天更新一次（第一次約 1–2 分鐘）"""
    cache = None
    if os.path.exists(SP500_CACHE):
        cache = pd.read_csv(SP500_CACHE)
        fresh = datetime.now().timestamp() - os.path.getmtime(SP500_CACHE) < SP500_CACHE_DAYS * 86400
        if fresh and cache["shares"].notna().mean() > 0.9:
            return cache
    try:
        df = sp500_constituents()
        with ThreadPoolExecutor(max_workers=8) as ex:
            df["shares"] = list(ex.map(_shares, df["symbol"]))
    except Exception:
        if cache is not None:   # 更新失敗就沿用舊快取
            return cache
        raise
    df.to_csv(SP500_CACHE, index=False)
    return df


def sp500_heatmap_data():
    uni = sp500_universe()
    px = last_two_closes(uni["symbol"])
    df = uni[uni["symbol"].isin(list(px))].copy()
    df["chg"] = [(px[s][0] / px[s][1] - 1) * 100 for s in df["symbol"]]
    df["mcap"] = [px[s][0] for s in df["symbol"]] * df["shares"]
    df["mcap"] = df["mcap"].fillna(df["mcap"].median() if df["mcap"].notna().any() else 1.0)
    df = df[df["mcap"] > 0].reset_index(drop=True)
    df.attrs["date"] = max(px[s][2] for s in df["symbol"])
    return df


def build_data_tasks():
    today = et_today()
    watch = {t.upper(): t.upper() for t in TICKERS}
    symbols = list({**MAIN_INDEXES, **OTHER_INDEXES, **MARKETS, **SECTOR_ETFS, **watch})
    start = today - timedelta(days=500)
    tasks = [Task("data", "quotes", "Yahoo Finance 行情", lambda: last_two_closes(symbols))]
    tasks += [Task("data", f"movers:{s}", f"Yahoo 排行 {s}", lambda s=s: screen_quotes(s))
              for s in ("day_gainers", "day_losers", "most_actives")]
    tasks.append(Task("data", "treasury", "財政部殖利率曲線", treasury_curve))
    tasks += [Task("data", f"fred:{sid}", f"FRED {sid}", lambda sid=sid: fred_series(sid, start))
              for sid in FRED_SERIES]
    if FRED_API_KEY:
        tasks.append(Task("data", "fred_calendar", "FRED 經濟數據行事曆", lambda: fred_calendar(today, CALENDAR_DAYS)))
    if FINNHUB_API_KEY:
        tasks.append(Task("data", "earnings:today", "Finnhub 今日財報", lambda: finnhub_earnings(today)))
        tasks.append(Task("data", "earnings:next", "Finnhub 下一交易日財報",
                          lambda: finnhub_earnings(next_weekday(today))))
    tasks.append(Task("data", "earnings:watch", "yfinance 自選股財報日", yf_watch_earnings))
    tasks.append(Task("data", "fear_greed", "CNN 恐懼與貪婪指數", fear_greed))
    if "S&P 500 熱力圖" in REPORT_ORDER:
        tasks.append(Task("data", "sp500", "S&P 500 成分股行情", sp500_heatmap_data))
    return tasks


def run_task(task):
    try:
        return task.fn(), None
    except Exception as e:
        return [], safe_err(e)
```

```python
# ---------- 報告各段落 ----------
from matplotlib.colors import LinearSegmentedColormap, Normalize
from matplotlib.figure import Figure
from matplotlib.patches import Rectangle


def change_of(s, q):
    """回傳 (顯示文字, 數值)；殖利率用 bp，其他用 %"""
    if s not in q:
        return "—", None
    last, prev, _ = q[s]
    if s in YIELD_SYMBOLS:
        v = (last - prev) * 100
        return f"{v:+.1f} bp", v
    v = (last / prev - 1) * 100
    return pct(v), v


def quote_table(mapping, q, label, sort=False):
    rows = []
    for s, name in mapping.items():
        text, v = change_of(s, q)
        rows.append(([name, s, num(q[s][0]) if s in q else "—", text], v))
    if sort:
        rows.sort(key=lambda r: -999 if r[1] is None else r[1], reverse=True)
    return md_table([label, "代號", "收盤", "漲跌"], [r for r, _ in rows])


def err_line(key, what):
    return f"- ❌ {what}無法取得：{ERRS[key]}" if key in ERRS else ""


def bp(a, b):
    return "—" if pd.isna(a) or pd.isna(b) else f"{(a - b) * 100:+.0f} bp"


def collect_news():
    cutoff = fetched_at - timedelta(hours=HOURS)
    tickers = list(dict.fromkeys(t.group for t in news_tasks if t.section == "ticker"))
    out = {"market": [], "official": [], "ticker": {t: [] for t in tickers}}
    for task, (items, err) in zip(news_tasks, news_results):
        fresh = [it for it in items if it.published and it.published >= cutoff]
        if task.section == "ticker":
            out["ticker"][task.group] += fresh
        else:
            out[task.section] += fresh
    out["market"] = dedup_sort(out["market"], MAX_MARKET)
    out["official"] = dedup_sort(out["official"], MAX_OFFICIAL)
    out["ticker"] = {t: dedup_sort(v, MAX_PER_TICKER) for t, v in out["ticker"].items()}
    return out


def earnings_note(t):
    for key, label in [("earnings:today", "今日"), ("earnings:next", "下一交易日")]:
        for d in D.get(key) or []:
            if d.get("symbol") == t:
                note = label + HOUR_LABEL.get(d.get("hour"), "")
                if key == "earnings:today" and d.get("epsActual") is not None:
                    note += f"｜EPS {num(d['epsActual'])} vs 預估 {num(d.get('epsEstimate'))}"
                return note
    w = (D.get("earnings:watch") or {}).get(t)
    return f"{w[0]:%m/%d}" if w else "—"


def earnings_table(items, reported):
    tracked = {t.upper() for t in TICKERS}
    items = sorted(items, key=lambda d: (d.get("symbol") not in tracked, -(d.get("revenueEstimate") or 0)))
    rows = []
    for d in items[:EARNINGS_TOP_N]:
        sym = ("★ " if d.get("symbol") in tracked else "") + d.get("symbol", "")
        when = HOUR_LABEL.get(d.get("hour"), d.get("hour") or "—")
        if reported:
            rows.append([sym, when, num(d.get("epsActual")), num(d.get("epsEstimate")),
                         billions(d.get("revenueActual")), billions(d.get("revenueEstimate"))])
        else:
            rows.append([sym, when, num(d.get("epsEstimate")), billions(d.get("revenueEstimate"))])
    if not rows:
        return "- （無）"
    headers = (["代號", "時段", "EPS 實際", "EPS 預估", "營收實際", "營收預估"] if reported
               else ["代號", "時段", "EPS 預估", "營收預估"])
    return md_table(headers, rows)


def fg_label(score):
    for limit, label in [(25, "極度恐懼"), (45, "恐懼"), (55, "中性"), (75, "貪婪"), (101, "極度貪婪")]:
        if score < limit:
            return label


# ----- 各段落 -----
def sec_indices():
    q = D.get("quotes", {})
    head = "、".join(f"{n} {change_of(s, q)[0]}" for s, n in MAIN_INDEXES.items() if s in q)
    lines = [f"**一句話**：{head or '（無資料）'}", "", quote_table(MAIN_INDEXES, q, "指數"), ""]
    others = "、".join(f"{n} {num(q[s][0])}（{change_of(s, q)[0]}）" for s, n in OTHER_INDEXES.items() if s in q)
    if others:
        lines.append(f"- 其他：{others}")
    latest = max((d for _, _, d in q.values()), default="—")
    lines += [f"- 資料日期：{latest}（Yahoo Finance，約 15 分鐘延遲）", err_line("quotes", "行情")]
    return "\n".join(lines).rstrip()


def sec_stocks():
    q = D.get("quotes", {})
    rows = [[t, num(q[t][0]) if t in q else "—", change_of(t, q)[0], earnings_note(t)] for t in NEWS["ticker"]]
    lines = [md_table(["個股", "收盤", "漲跌", "財報"], rows), ""]
    for t, items in NEWS["ticker"].items():
        lines += [f"### {t}　{change_of(t, q)[0]}", "", news_list(items, "- （無新聞）"), ""]
    return "\n".join(lines).rstrip()


def sec_focus_news():
    top = [it for it in NEWS["market"] if is_key(it.title)][:MAX_HIGHLIGHTS]
    rest = [it for it in NEWS["market"] if it not in top]
    return f"### ★ 重點（標題含關鍵字）\n\n{news_list(top)}\n\n### 其他市場新聞\n\n{news_list(rest)}"


def sec_sectors():
    q = D.get("quotes", {})
    ranked = sorted((s for s in SECTOR_ETFS if s in q), key=lambda s: change_of(s, q)[1], reverse=True)
    lines = []
    if ranked:
        best, worst = ranked[0], ranked[-1]
        lines += [f"- 最強：{SECTOR_ETFS[best]} {change_of(best, q)[0]}｜最弱：{SECTOR_ETFS[worst]} {change_of(worst, q)[0]}", ""]
    lines.append(quote_table(SECTOR_ETFS, q, "類股（SPDR ETF）", sort=True))
    return "\n".join(lines)


def sec_movers():
    lines = []
    for key, title in [("day_gainers", "漲幅排行"), ("day_losers", "跌幅排行"), ("most_actives", "成交最活躍")]:
        lines += [f"### {title}", ""]
        quotes = D.get(f"movers:{key}")
        if quotes is None:
            lines.append(err_line(f"movers:{key}", title) or "- （無資料）")
        else:
            rows = [[x.get("symbol", ""), clean_text(x.get("shortName") or x.get("longName") or "", 28),
                     num(x.get("regularMarketPrice")), pct(x.get("regularMarketChangePercent")),
                     f"{x['marketCap'] / 1e9:,.1f}B" if x.get("marketCap") else "—"] for x in quotes]
            lines.append(md_table(["代號", "名稱", "價格", "漲跌", "市值"], rows) if rows else "- （無資料）")
        lines.append("")
    return "\n".join(lines) + "> Yahoo Finance 預設篩選條件。"


def sec_rates_fx():
    lines = ["### 美國公債殖利率（財政部）", ""]
    curve = D.get("treasury")
    if curve is not None and len(curve) >= 2:
        cur, prev = curve.iloc[0], curve.iloc[1]
        rows = [[t, num(cur[t]), bp(cur[t], prev[t])] for t in TENORS if t in curve]
        lines += [f"資料日期：{cur['Date']:%Y-%m-%d}（與 {prev['Date']:%m/%d} 比較）", "",
                  md_table(["天期", "殖利率 %", "日變動"], rows), "",
                  f"- 10 年 − 2 年利差：{(cur['10 Yr'] - cur['2 Yr']) * 100:+.0f} bp"
                  f"（前一日 {(prev['10 Yr'] - prev['2 Yr']) * 100:+.0f} bp）"]
        if et_today().weekday() < 5 and cur["Date"].date() < et_today():
            lines.append("- 財政部當日數據尚未公布，以上為前一交易日；即時殖利率見下表。")
    else:
        lines.append(err_line("treasury", "財政部殖利率") or "- （無資料）")
    lines += ["", "### 即時利率、美元與商品（Yahoo Finance）", "",
              quote_table(MARKETS, D.get("quotes", {}), "項目")]
    return "\n".join(lines)


def sec_macro():
    rows = []
    for sid, name in FRED_SERIES.items():
        s = D.get(f"fred:{sid}")
        if s is None or len(s) < 2:
            rows.append([name, sid, "—", "—", "❌ " + ERRS.get(f"fred:{sid}", "資料不足")[:60]])
        elif sid in YOY_SERIES:
            yoy = (s["value"] / s["value"].shift(12) - 1) * 100
            rows.append([name, sid, f"{yoy.iloc[-1]:.2f}", f"{yoy.iloc[-1] - yoy.iloc[-2]:+.2f}", s["date"].iloc[-1]])
        else:
            rows.append([name, sid, num(s["value"].iloc[-1]),
                         f"{s['value'].iloc[-1] - s['value'].iloc[-2]:+,.2f}", s["date"].iloc[-1]])
    lines = ["### 重要指標（FRED）", "", md_table(["項目", "代號", "最新", "變動", "資料日期"], rows), "",
             f"### 未來 {CALENDAR_DAYS} 天經濟數據（美東日期）", ""]
    if "fred_calendar" in D:
        lines.append(md_table(["日期", "發布項目"], D["fred_calendar"]) if D["fred_calendar"] else "- （無）")
    else:
        lines.append(err_line("fred_calendar", "經濟數據行事曆") or "- 在 .env 加入 FRED_API_KEY（免費申請）即可顯示。")
    lines += ["", "> 預期值與公布值請對照 Investing.com 經濟日曆。", "",
              "### 聯準會公告", "", news_list(NEWS["official"], "- （時間範圍內無新公告）")]
    return "\n".join(lines)


def sec_earnings():
    today, nxt = et_today(), next_weekday(et_today())
    if "earnings:today" in D or "earnings:today" in ERRS:
        lines = [f"### 今日（美東 {today:%m/%d}）", "",
                 earnings_table(D["earnings:today"], True) if "earnings:today" in D
                 else err_line("earnings:today", "今日財報"),
                 "", f"### 下一交易日（{nxt:%m/%d}）預告", "",
                 earnings_table(D["earnings:next"], False) if "earnings:next" in D
                 else err_line("earnings:next", "財報預告"),
                 "", "> ★ 為自選股；盤後公布的財報，實際數字可能稍晚才更新。"]
    else:
        watch = D.get("earnings:watch") or {}
        rows = [[t, f"{d:%Y-%m-%d}", num(e), billions(r)]
                for t, (d, e, r) in sorted(watch.items(), key=lambda kv: kv[1][0])]
        lines = ["### 自選股未來 14 天財報（yfinance）", "",
                 md_table(["代號", "日期", "EPS 預估", "營收預估"], rows) if rows
                 else (err_line("earnings:watch", "財報日") or "- （無）"),
                 "", "> 設定 FINNHUB_API_KEY 可顯示全市場財報行事曆與實際 vs 預估。"]
    return "\n".join(lines)


def sec_sentiment():
    lines = []
    fg = D.get("fear_greed")
    if fg:
        rows = [["目前", f"{fg['score']:.0f}", fg_label(fg["score"])]]
        for k, label in [("previous_close", "前一日"), ("previous_1_week", "一週前"),
                         ("previous_1_month", "一個月前"), ("previous_1_year", "一年前")]:
            if fg.get(k) is not None:
                rows.append([label, f"{fg[k]:.0f}", fg_label(fg[k])])
        updated = datetime.fromisoformat(fg["timestamp"]).astimezone(TZ_LOCAL)
        lines += ["### CNN 恐懼與貪婪指數", "", md_table(["時間", "分數（0–100）", "狀態"], rows), "",
                  f"- CNN 更新時間：台北 {updated:%m/%d %H:%M}"]
    else:
        lines.append(err_line("fear_greed", "CNN 恐懼與貪婪指數") or "- （無資料）")
    q = D.get("quotes", {})
    if "^VIX" in q:
        lines.append(f"- VIX：{q['^VIX'][0]:.2f}（{change_of('^VIX', q)[0]}）")
    return "\n".join(lines)


def squarify(values, x, y, w, h):
    """把由大到小排序的正數切成面積成比例的矩形（squarified treemap），回傳 [(x, y, w, h), ...]"""
    total = float(sum(values))
    vals = [v * w * h / total for v in values]
    rects = []

    def worst(row, side):
        s = sum(row)
        return max(max(side * side * r / (s * s), (s * s) / (side * side * r)) for r in row)

    while vals:
        side = min(w, h)
        row = [vals.pop(0)]
        while vals and worst(row + [vals[0]], side) <= worst(row, side):
            row.append(vals.pop(0))
        thick, pos = sum(row) / side, 0.0
        for v in row:
            length = v / thick
            rects.append((x, y + pos, thick, length) if w >= h else (x + pos, y, length, thick))
            pos += length
        if w >= h:
            x, w = x + thick, w - thick
        else:
            y, h = y + thick, h - thick
    return rects


def draw_heatmap(df, path, title):
    """S&P 500 熱力圖：類股分區、方塊大小 = 市值、顏色 = 漲跌幅"""
    W, H, bg = 1600, 900, "#262931"
    colors = ["#30cc5a", "#414554", "#f63538"] if HEATMAP_RED_UP else ["#f63538", "#414554", "#30cc5a"]
    cmap = LinearSegmentedColormap.from_list("heat", colors)
    norm = Normalize(-HEATMAP_RANGE, HEATMAP_RANGE)

    fig = Figure(figsize=(16, 9.6), dpi=110, facecolor=bg)
    ax = fig.add_axes([0.005, 0.005, 0.99, 0.935])
    ax.set_xlim(0, W)
    ax.set_ylim(H, 0)
    ax.axis("off")
    fig.text(0.01, 0.97, title, color="white", fontsize=15, weight="bold", va="center")

    sectors = df.groupby("sector")["mcap"].sum().sort_values(ascending=False)
    for (sx, sy, sw, sh), (sector, total) in zip(squarify(sectors.values, 0, 0, W, H), sectors.items()):
        sub = df[df["sector"] == sector].sort_values("mcap", ascending=False)
        head = min(18, sh * 0.15)
        sec_chg = (sub["chg"] * sub["mcap"]).sum() / total
        label, maxc = f"{sector.upper()} {sec_chg:+.2f}%", int(sw / 6.7)
        if len(label) > maxc:
            label = f"{sector.upper()[:maxc - 9]}… {sec_chg:+.2f}%" if maxc > 13 else ""
        if label and head >= 9:
            ax.text(sx + 4, sy + head / 2, label, color="#d9dbe1", fontsize=8, weight="bold", va="center")
        tiles = squarify(sub["mcap"].values, sx + 1, sy + head, sw - 2, sh - head - 1)
        for (x, y, w, h), r in zip(tiles, sub.itertuples()):
            ax.add_patch(Rectangle((x, y), w, h, facecolor=cmap(norm(r.chg)), edgecolor=bg, linewidth=0.7))
            fs = min(22, (w * h) ** 0.5 / 5, w / (max(len(r.symbol), 6) * 1.05), h * 0.28)
            if fs >= 5.5:
                ax.text(x + w / 2, y + h / 2, f"{r.symbol}\n{r.chg:+.2f}%", color="white", fontsize=fs,
                        weight="bold", ha="center", va="center", linespacing=1.15)
            elif min(w / (len(r.symbol) * 1.05), h * 0.55) >= 5:
                ax.text(x + w / 2, y + h / 2, r.symbol, color="white", fontsize=5, ha="center", va="center")
        ax.add_patch(Rectangle((sx, sy), sw, sh, fill=False, edgecolor=bg, linewidth=2.5))
    fig.savefig(path, facecolor=bg)


def sec_heatmap():
    df = D.get("sp500")
    if df is None or df.empty:
        return err_line("sp500", "S&P 500 熱力圖資料") or "- （無資料）"
    path = f"sp500_heatmap_{fetched_at.astimezone(TZ_LOCAL):%Y%m%d_%H%M}.png"
    draw_heatmap(df, path, f"S&P 500  |  {df.attrs.get('date', '')}  |  size: market cap   color: daily change")
    with open(path, "rb") as f:
        IMAGES[path] = base64.b64encode(f.read()).decode()
    contrib = df["mcap"] / df["mcap"].sum() * df["chg"]
    ranked = df.assign(contrib=contrib).sort_values("contrib")
    names = lambda rows: "、".join(f"{r.symbol} {pct(r.chg)}" for r in rows.itertuples())
    return "\n".join([
        f"![S&P 500 熱力圖]({path})", "",
        f"- 上漲 {(df['chg'] > 0).sum()} 檔、下跌 {(df['chg'] < 0).sum()} 檔（共 {len(df)} 檔）｜市值加權漲跌 {pct(contrib.sum())}",
        f"- 推升指數最多：{names(ranked.tail(5).iloc[::-1])}",
        f"- 拖累指數最多：{names(ranked.head(5))}",
        f"- 方塊大小為市值（股價 × 流通股數，近似值），顏色為當日漲跌，"
        f"{'紅漲綠跌' if HEATMAP_RED_UP else '綠漲紅跌'}，±{HEATMAP_RANGE:g}% 為最深色",
    ])


def sec_status():
    cutoff = fetched_at - timedelta(hours=HOURS)
    rows = []
    for task, (value, err) in zip(all_tasks, results):
        if err:
            rows.append([task.label, 0, "—", "❌ " + err])
        elif task.section == "data":
            n = 1 if task.group == "fear_greed" else (len(value) if hasattr(value, "__len__") else 1)
            rows.append([task.label, n, "—", "✅"])
        else:
            fresh = sum(1 for it in value if it.published and it.published >= cutoff)
            rows.append([task.label, len(value), fresh, "✅"])
    rows.sort(key=lambda r: not r[3].startswith("❌"))  # 失敗的排最前面
    notes = [f"> 未設定 {k}，已略過相關來源。" for k, v in
             [("FINNHUB_API_KEY", FINNHUB_API_KEY), ("FRED_API_KEY", FRED_API_KEY), ("SEC_CONTACT", SEC_CONTACT)] if not v]
    return md_table(["來源", "筆數", "時間範圍內", "狀態"], rows) + ("\n\n" + "\n".join(notes) if notes else "")


SECTION_BUILDERS = {
    "四大指數": sec_indices,
    "個股資訊": sec_stocks,
    "市場焦點新聞": sec_focus_news,
    "S&P 500 熱力圖": sec_heatmap,
    "類股表現": sec_sectors,
    "漲跌排行": sec_movers,
    "債市、匯市與商品": sec_rates_fx,
    "經濟數據與聯準會": sec_macro,
    "財報行事曆": sec_earnings,
    "市場情緒": sec_sentiment,
    "資料來源狀況": sec_status,
}
print("函式載入完成")
```
